```python
def generate_candidate(career):
    profile = career_profiles[career]

    max_skills = len(profile["skills"])
    num_skills = random.randint(3, max_skills)

    selected_skills = random.sample(profile["skills"], num_skills)

    noise_count = random.randint(0, 2)
    noise_skills = random.sample(common_skills, noise_count)

    all_skills = list(set(selected_skills + noise_skills))

    education = random.choice(profile["education"])

    mixed_interests = profile["interests"] + [
        "technology",
        "innovation",
        "problem solving"
    ]

    interest = random.choice(mixed_interests)

    experience = random.randint(0, 8)

    return {
        "skills": " ".join(all_skills),
        "education": education,
        "interests": interest,
        "experience": experience,
        "career": career
    }
```

```python
generate_candidate("Data Scientist")
```

```python
generate_candidate("AI Engineer")
```

```python
dataset = []

for career in career_profiles:
    for _ in range(100):
        dataset.append(generate_candidate(career))

df = pd.DataFrame(dataset)
```
